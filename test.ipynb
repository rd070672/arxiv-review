```python
以下のような簡単な例題が考えられます。

- フィボナッチ数列
   - 入力 `n` に対して、最初の `n` 個のフィボナッチ数を返す関数を作る。
   - 例：`n=7` → `[1, 1, 2, 3, 5, 8, 13]`

- 回文判定
   - 文字列を入力として受け取り、前後から読んでも同じかどうかを判定する関数を作る。
   - 例：`"radar"` → `True`、`"python"` → `False`

- リストの要素操作
   - 整数のリストを受け取って、偶数だけ抽出し、それを二乗した新しいリストを返す。
   - `list` と `for`、あるいはリスト内包表記の練習になる。
  - 例：`[1, 2, 3, 4, 5]` → `[4, 16]`

- 辞書を使った頻度カウント
   - 文字列やリストを入力として、各要素の出現回数を辞書で集計する関数を作る。
   - 例：`"banana"` → `{'b': 1, 'a': 3, 'n': 2}`

- ファイル読み込みと集計
   - テキストファイルを読み込み、行数・単語数・文字数を数える簡単なスクリプトを作る。
   - 例：`sample.txt` の内容を読み取り、`lines=10`, `words=45`, `chars=250` のように出力する。
```

```python
def fibonacci(n):
   result = []
   a, b = 1, 1
   for _ in range(n):
      result.append(a)
      a, b = b, a + b + 1
   return result

# 例
print(fibonacci(7))  # [1, 1, 2, 3, 5, 8, 13]
```
